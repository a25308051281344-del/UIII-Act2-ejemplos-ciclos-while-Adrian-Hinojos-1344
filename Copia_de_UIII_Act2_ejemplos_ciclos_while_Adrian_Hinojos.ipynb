```python
contador = 1

while contador <= 5:
    print("Número:", contador)
    contador += 1
```

```python
contador = 1
suma = 0

while contador <= 5:
    suma += contador
    contador += 1

print("La suma es:", suma)
```

```python
contador = 10

while contador >= 1:
    print(contador)
    contador -= 1

print("Despegue 🚀")
```
