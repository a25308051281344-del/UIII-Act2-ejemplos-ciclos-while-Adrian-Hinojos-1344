```python
texto = "inteligencia artificial"
indice = 0
contador = 0

while indice < len(texto):
    if texto[indice] in "aeiou":
        contador += 1
    indice += 1

print("Vocales:", contador)
```

```python
ahorro = 0

while ahorro < 500:
    deposito = int(input("Deposita dinero: "))
    ahorro += deposito
    print("Total ahorrado:", ahorro)

print("Meta alcanzada 🎉")
```

```python
contador = 1

while contador <= 5:
    print("Número:", contador)
    contador += 1
```

```python
contador = 1
suma = 0

while contador <= 5:
    suma += contador
    contador += 1

print("La suma es:", suma)
```

```python
contador = 10

while contador >= 1:
    print(contador)
    contador -= 1

print("Despegue 🚀")
```
